```python
matches_subset_df = matches.select("match_id", "season", "scheduled_overs", "gender", "team_type",\
    "city", "venue", "team1", "team2", "toss_decision", "outcome_result", "outcome_winner", \
        "win_by_runs", "win_by_wickets", "player_of_match")
```

```python
matches_subset_df.filter(
    # pl.col("city") == "Delhi",
    # pl.col("gender") == "male",
    # pl.col("scheduled_overs") == 20,
    (
        ((pl.col("team1") == "India") & (pl.col("team2") == "Afghanistan")) |
        ((pl.col("team1") == "Afghanistan") & (pl.col("team2") == "India"))
    )
)
```

```python
matches_subset_df.filter(
    ((pl.col("team1")=="India")|(pl.col("team2")=="India")),
    pl.col("gender") == "male"
    )\
    .get_column("team1").unique().to_list()
```

```python
matches_subset_df.filter(
    ((pl.col("team1")=="India")|(pl.col("team2")=="India")),
    pl.col("gender") == "male"
    )\
    .get_column("team2").unique().to_list()
```

```python
matches_subset_df.filter(
    ((pl.col("team1")=="India")|(pl.col("team2")=="India")),
    pl.col("gender") == "male",
    pl.col("city") == "Delhi"
    )
```

```python
print(sorted(matches_subset_df.filter(
    pl.col("gender") == "male"
    )\
    .get_column("team1").unique().to_list()))
```
